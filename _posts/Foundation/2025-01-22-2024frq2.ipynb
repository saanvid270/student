```python
public class Main {
    // Instance variables to track the game state
    private String nameOfTeam1;
    private String nameOfTeam2;
    private int pointsForTeam1;
    private int pointsForTeam2;
    private int currentTurn;  // 1 means team1's turn, 2 means team2's turn
    
    // Constructor initializes a new game with two team names
    public Main(String firstTeamName, String secondTeamName) {
        nameOfTeam1 = firstTeamName;
        nameOfTeam2 = secondTeamName;
        pointsForTeam1 = 0;
        pointsForTeam2 = 0;
        currentTurn = 1;  // Team 1 always starts
    }
    
    // Helper method that returns the name of whichever team has the current turn
    private String getTeamWithTurn() {
        if (currentTurn == 1) {
            return nameOfTeam1;
        } else {
            return nameOfTeam2;
        }
    }
    
    // Helper method that switches the turn to the other team
    private void changeTurn() {
        if (currentTurn == 1) {
            currentTurn = 2;
        } else {
            currentTurn = 1;
        }
    }
    
    // Main method that processes a play in the game
    public void recordPlay(int points) {
        if (points > 0) {
            // If points are positive, add them to the team with the current turn
            if (currentTurn == 1) {
                pointsForTeam1 += points;
            } else {
                pointsForTeam2 += points;
            }
        } else {
            // If points are 0, the team failed so switch turns
            changeTurn();
        }
    }
    
    // Returns the current score and active team as a formatted string
    public String getScore() {
        String activeTeamName = getTeamWithTurn();
        return pointsForTeam1 + "-" + pointsForTeam2 + "-" + activeTeamName;
    }
    
    // Main method to test the Scoreboard functionality
    public static void main(String[] args) {
        System.out.println("=== Testing Scoreboard ===\n");
        
        // Create a new game between Red and Blue
        Main game = new Main("Red", "Blue");
        
        System.out.println("Initial state: " + game.getScore());
        
        game.recordPlay(1);
        System.out.println("Red scores 1: " + game.getScore());
        
        game.recordPlay(0);
        System.out.println("Red fails (Blue's turn): " + game.getScore());
        
        game.recordPlay(3);
        System.out.println("Blue scores 3: " + game.getScore());
        
        game.recordPlay(1);
        game.recordPlay(0);
        System.out.println("Blue scores 1 then fails: " + game.getScore());
        
        game.recordPlay(0);
        game.recordPlay(4);
        game.recordPlay(0);
        System.out.println("Multiple plays: " + game.getScore());
        
        System.out.println("\n=== Testing Multiple Games ===\n");
        
        // Create a second independent game
        Main match = new Main("Lions", "Tigers");
        System.out.println("New game: " + match.getScore());
        System.out.println("Original game still: " + game.getScore());
    }
}
```

## Code Explanation

**Constructor: public Main(String firstTeamName, String secondTeamName)**
This special method runs automatically when you create a new game object using new Main("Red", "Blue"). It takes two team names as parameters and initializes all the instance variables. Both teams start with 0 points, and team 1 is always set as the starting team by setting currentTurn = 1. This sets up a fresh game ready to track plays.

**Helper Method: getTeamWithTurn()**
This private method figures out which team currently has possession. It checks the value of currentTurn and returns the appropriate team name. If currentTurn is 1, it returns nameOfTeam1; otherwise it returns nameOfTeam2. This method is helpful because multiple other methods need to know whose turn it is.

**Helper Method: changeTurn()**
This private method switches possession between teams. It uses a simple if-else statement to toggle currentTurn between 1 and 2. If the current turn is 1, it changes to 2. If it's 2, it changes back to 1. This gets called whenever a team fails to score.

**Main Game Method: recordPlay(int points)**
This is the core method that processes what happens during each play in the game. It takes an integer parameter representing points scored. If the points are greater than 0, those points get added to whichever team has the current turn. If points equal 0, that means the team failed their attempt, so the method calls changeTurn() to switch possession to the other team without adding any points.

**Score Method: getScore()**
This method generates and returns a String showing the current game state. It creates a formatted string in the pattern "score1-score2-activeTeam" by concatenating the two team scores with hyphens and the name of the team that currently has the turn. It calls getTeamWithTurn() to determine which team name should appear at the end.

**Static Main Method: public static void main(String[] args)**
This is the special entry point where the program begins execution. It's static, meaning it belongs to the class itself rather than to individual game objects. Inside this method, we create test game objects and simulate various plays to verify the scoreboard works correctly. It prints output to the console showing how the game state changes after each play.
